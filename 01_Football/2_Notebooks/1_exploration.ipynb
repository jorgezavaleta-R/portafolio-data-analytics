```python
# Cruce directo de 'type' contra 'competencia'
events = pd.crosstab(df['type'], df['competencia']).reset_index()

# Renombramos el nombre del índice de columnas si es necesario
events.columns.name = None

events['Total'] = events['wc_18'] + events['wc_22']

events = events.sort_values(by='Total', ascending=False).reset_index(drop=True)

events
```
