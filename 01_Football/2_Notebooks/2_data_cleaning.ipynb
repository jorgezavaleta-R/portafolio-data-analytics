```python
import pandas as pd
import numpy as np
import ast
```

```python
file_path = '/Users/jzavaleta/Desktop/portafolio-data-analytics/01_Football/1_Data/wc_Mexico.csv'
df = pd.read_csv(file_path)
```

```python
def extraer_coordenadas(val):
    # Si es nulo o 'Missing value', regresamos nulos para X e Y
    if pd.isna(val) or val is None or val == 'Missing value':
        return np.nan, np.nan
    
    # Si es un string tipo '[60.0, 40.0]', lo convertimos a lista real
    if isinstance(val, str):
        try:
            val = ast.literal_eval(val)
        except (ValueError, SyntaxError):
            return np.nan, np.nan
            
    # Extraemos el primer y segundo elemento numérico
    if isinstance(val, (list, tuple)) and len(val) >= 2:
        return val[0], val[1]
        
    return np.nan, np.nan

# Aplicamos la función a la columna location
df[['x_start', 'y_start']] = df['location'].apply(extraer_coordenadas).tolist()
df[['x_end', 'y_end']] = df['pass_end_location'].apply(extraer_coordenadas).tolist()

# Mostramos el resultado
df[['location', 'x_start', 'y_start','pass_end_location','x_end', 'y_end']].head(10)
```
